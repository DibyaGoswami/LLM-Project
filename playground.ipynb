```python
# We initialize the embedding layer by giving it dimensions
vocab_size = 10000
embedding_dim = 100
embedding = nn.Embedding(vocab_size, embedding_dim)

# Then we create some input indices
input_indices = torch.LongTensor([1, 5, 3, 2])

# Apply the embedding layer onto the input indices
embedded_output = embedding(input_indices)

# output will be a tensor of shape (4, 100) where 4 is the number of indices
# and 100 is the dimensionality of the embedding vectors
print(embedded_output.shape)
```
